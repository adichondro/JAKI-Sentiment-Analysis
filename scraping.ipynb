# Scraping Ulasan Aplikasi JAKI dari Google Play Store
Notebook ini digunakan untuk mengumpulkan dataset ulasan aplikasi JAKI dari Google Play Store secara otomatis.

```python
import pandas as pd
from google_play_scraper import reviews_all, Sort

# Proses akuisisi data ulasan aplikasi JAKI dari Google Play Store
scrapreview = reviews_all(
    'id.go.jakarta.smartcity.jaki',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=12000
)

app_reviews_df = pd.DataFrame(scrapreview)
print(f"Total data ulasan yang berhasil dikumpulkan: {len(app_reviews_df)}")
```

```python
# Simpan seluruh dataset mentah hasil scraping ke CSV
app_reviews_df.to_csv('ulasan_aplikasi_jaki_raw.csv', index=False)

jumlah_ulasan, jumlah_kolom = app_reviews_df.shape
print(f"Dataset berhasil disimpan! Total ulasan: {jumlah_ulasan} baris, {jumlah_kolom} kolom.")
```
